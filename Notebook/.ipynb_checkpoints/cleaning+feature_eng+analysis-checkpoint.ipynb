# CarDekho New Cars — Web Scraping, Feature Engineering & EDA

This project scrapes live listing data for new cars from CarDekho.com using Selenium, 
then cleans and engineers features from the raw scraped fields to enable price-segment 
and specification analysis across 293 car models currently listed on the platform.

**Workflow:**
1. Scrape car name, rating, price range, mileage, and engine displacement using Selenium 
   (handles CarDekho's "Load More" pagination dynamically until all 293 listings are loaded)
2. Clean and engineer features from raw text fields (price ranges, unit conversion, brand extraction)
3. Handle missing values using segment-aware imputation
4. Perform univariate and bivariate exploratory analysis

Before engineering new features, the raw scrape returns **293 rows, 5 columns**, with 
notable missingness already visible: `rating` is missing in 8 rows, `mileage` in 115 
rows, and `engine` in 65 rows — mileage in particular is often absent for electric 
vehicles, which don't have a fuel-based mileage figure. This missingness pattern is 
what later justifies the segment-aware imputation approach instead of a simple global fill.

## FEATURE ENGINEERING

```python
def clean_price_split(price_str):
    if pd.isna(price_str):
        return None,None
        
    clean_str = str(price_str).split('*')[0].replace('Rs.','').strip()

    unit = 'Cr' if 'Cr' in clean_str else 'Lakh'
    clean_str = clean_str.replace('Lakh','').replace('Cr','').strip()

    if '-' in clean_str:
        min_val, max_val = map(float, clean_str.split('-'))
    else :
        min_val = max_val = float(clean_str)

    if unit == "Cr":
        min_val *= 100
        max_val *= 100
    return pd.Series([min_val, max_val])
    
df[["price_min_lakh","price_max_lakh"]] = df["price"].apply(clean_price_split)
```

`price` on CarDekho is scraped as a single text blob like `"Rs.11.99 - 19.19 Lakh*Get 
On-Road Price\n*Ex-Showroom..."`, mixing the actual price range with disclaimer text, 
and inconsistently using **Lakh** for most cars but **Cr (Crore)** for high-end models 
(e.g. Rolls-Royce, Bentley). **clean_price_split()** strips the disclaimer text, detects 
the unit, and normalizes everything to Lakh (1 Cr = 100 Lakh) so all 293 rows are 
directly comparable on one numeric scale.

```python
df["company"] = df["car_name"].str.split().str[0]
df['company'].info()
```

Brand is extracted from the first word of `car_name`. This works for most rows, but 
produces incorrect brand values for multi-word brand names — `"Range Rover Velar"` 
becomes company `"Range"`, `"Land Cruiser"` becomes `"Land"`, and `"Aston Martin"` 
becomes `"Aston"`. The unique company list (44 total) , Understood about the company and replace multi worded companies into its orginal form manually
(Range Rover, Land Rover, Aston Martin, Rolls-Royce) if brand-level analysis is a focus area.

```python
brand_fix = {"Range":"Range Rover Velar","Aston":"Aston Martin","Land":"Land Rover"}
df["company"]= df["company"].astype(str)
df["company"]= df["company"].str.replace(brand_fix)
df["company"]= df["company"].astype(object)
```

```python
print(df["company"].unique())
```

```python
df['price_avg_lakh'] = (df['price_min_lakh'] + df['price_max_lakh']) / 2
```

```python
def price_segment(price_avg_lakh):
    if price_avg_lakh < 6:
        return "Budget"
    elif price_avg_lakh >= 6 and price_avg_lakh <= 10:
        return "Lower Mid-Range"
    elif price_avg_lakh > 10 and price_avg_lakh <= 20:
        return "Upper Mid-Range"
    elif price_avg_lakh > 20 and price_avg_lakh <= 50:
        return "Premium & Executive"
    else:
        return "Luxury & High-End"



df["price_segment"] = df["price_avg_lakh"].apply(price_segment)
```

```python
len(df[df["car_name"].str.contains("Electric|EV|XEV",case = False,na=False)]) 
```

```python
df["is_electric"] =   np.where(df["car_name"].str.contains("Electric|EV|XEV",case = False,na=False),1,0)
```

## CLEANING

```python
df.nunique()
```

```python
df = df.drop(columns=["price"])
```

```python
df = df.rename(columns = {"mileage":"mileage_km_per_liter","engine":"engine_cc"})
```

```python
df["engine_cc"] = df["engine_cc"].str.replace(" cc","")
```

```python
df["mileage_km_per_liter"] = df["mileage_km_per_liter"].str.replace(" kmpl","")
```

```python
df.isna().sum()
```

```python
df["engine_cc"] = pd.to_numeric(df["engine_cc"], errors = "coerce")
df["rating"] = pd.to_numeric(df["rating"], errors = "coerce")
df["mileage_km_per_liter"] = pd.to_numeric(df["mileage_km_per_liter"], errors = "coerce")
df["is_electric"] = pd.to_numeric(df["is_electric"], errors = "coerce")
```

## Handling missing value

```python
df["rating"] = df["rating"].fillna(df["rating"].median())
```

```python
df.loc[df["is_electric"] == 1,"engine_cc"] = 0
df.loc[df["is_electric"] == 1,"mileage_km_per_liter"] = 0
```

```python
df["mileage_km_per_liter"] = df["mileage_km_per_liter"].fillna(df.groupby("price_segment")["mileage_km_per_liter"].transform("median"))
df["engine_cc"] = df["engine_cc"].fillna(df.groupby("price_segment")["engine_cc"].transform("median"))
```

```python
df.isna().sum()
```

## UNI VARIATE ANALYSIS

```python
company_order = df["company"].value_counts(ascending=True).index
```

```python
plt.figure(figsize=(8,4))
sns.histplot(data=df,y=df["price_segment"])
plt.title("Distribution of car price segment")
plt.savefig("price_Segment.png",dpi=300,bbox_inches="tight")
plt.tight_layout()
plt.show()
```

**Most listings fall in the Luxury & High-End segments, showing the site's catalog leans toward mid-to-high price cars rather than budget options.**

```python
plt.figure(figsize=(10,12))
sns.countplot(data=df,y=df["company"],order=company_order)
plt.title("companies by freequency",fontsize=12)
plt.savefig("companies_freq.png",dpi=300,bbox_inches="tight")
plt.tight_layout()
plt.show()
```

**Mercedes-Benz dominate in the new cars list followed by BMW and also main stream companies like Maruti,Tata,Mahindra also hold some place in top 5**

```python
sns.boxplot(data=df,y="engine_cc")
plt.title("Engine cc boxplot")
plt.savefig("engine_cc.png")
plt.show()
```

**IQR METHOD FOR OUTLIER DETECTION**

```python
Q1 = df["engine_cc"].quantile(0.25)
Q3 = df["engine_cc"].quantile(0.75)

IQR = Q3-Q1

lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR

outlier_cc = df[(df["engine_cc"]<lower_bound)|(df["engine_cc"]>upper_bound)]
outlier_cc
```

**Engine size is tightly clustered for most cars, with a few extreme outliers from ultra-luxury brands pulling the range far to the right.
All 7 outliers belong to ultra-luxury brands (Bentley, Rolls-Royce, Ferrari) — these are genuine high-end vehicles, not data errors.**
